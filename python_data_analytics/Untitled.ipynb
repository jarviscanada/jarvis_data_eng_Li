```python
import pandas as pd
import numpy as np

data = {
"OrderID": [101, 102, 103, 104, 105, 106, 107, 108],
"Customer": ["Alice", "Bob", "Alice", "David", "Eva", "Bob", "Frank", "Eva"],
"Product": ["Laptop", "Phone", "Mouse", "Laptop", "Tablet", "Phone", "Tablet", "Mouse"],
"Category": ["Electronics", "Electronics", "Accessories", "Electronics", "Electronics", "Electronics", "Electronics", "Accessories"],
"Quantity": [1, 2, 3, 1, 2, 1, 1, 4],
"Price": [1200, 800, 25, 1200, 600, 800, 600, np.nan]
}

df = pd.DataFrame(data)
```

```python
df["Price"] = df["Price"].fillna(df.groupby("Product")["Price"].transform("mean"))
df
```

```python
df["High_Value_Order"] = np.where(df["TotalRevenue"] > 1500, "Yes", "No")
df
```

```python
customer_spend = df.groupby("Customer")["TotalRevenue"].sum()
```
