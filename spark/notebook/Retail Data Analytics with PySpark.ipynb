# Retail Data Wrangling and Analytics

```python
# Import modules 
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import datetime as dt
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import Window
from pyspark.ml.feature import QuantileDiscretizer
from pyspark.sql.functions import col,concat_ws,when
```

```python
spark = SparkSession.builder.appName("Retail Data Analytics").config("config.option", "value").getOrCreate()
```

```python
retail_df = spark.sql("SELECT * FROM hive_metastore.default.retail")
display(retail_df)
```

# Total Invoice Amount Distribution

```python
retail_df = retail_df.withColumn("total", F.col("quantity") * F.col("unit_price"))
invoice_amount = retail_df.groupby('invoice_no').agg(F.sum("total").alias("total"))
invoice_amount = invoice_amount.filter(F.col("total") > 0).sort(F.col("total").desc())
display(invoice_amount)
```

# Monthly Placed and Canceled Orders

```python
retail_df = retail_df.withColumn("yyyymm", F.date_format("invoice_date", "yyyyMM"))
retail_df = retail_df.withColumn("is_cancelled", F.col("invoice_no").cast("string").startswith("C"))
total_invoices = retail_df.groupby("yyyymm").agg(F.countDistinct("invoice_no").alias("Placement")).orderBy("yyyymm")
cancelled_invoices = retail_df[retail_df["is_cancelled"]].groupby("yyyymm").agg(F.countDistinct("invoice_no").alias("Cancellation")).orderBy("yyyymm")
```

```python
display(cancelled_invoices)
```

```python
merged_invoice = total_invoices.join(cancelled_invoices, on='yyyymm', how='left').orderBy("yyyymm", ascending=True)
display(merged_invoice)
```

# Monthly Sales

```python
total_amount=retail_df.groupby("yyyymm").agg(F.sum("total").alias("total")).orderBy("yyyymm")
display(total_amount)
```

# Monthly Sales Growth


```python
windowSpec = Window.orderBy("yyyymm")
lag = F.lag("total").over(windowSpec)
total_amount = total_amount.withColumn("growth", F.round(((F.col("total") - lag) / lag) * 100, 2))
display(total_amount["yyyymm","growth"])
```

# Monthly Active Users

```python
active_users = retail_df.groupby("yyyymm").agg(F.countDistinct("customer_id").alias("Active Users")).orderBy("yyyymm")
display(active_users)
```

# New and Existing Users



```python
users = retail_df.groupby("customer_id").agg(F.min("yyyymm").alias("users"))
retail_df = retail_df.join(users, on="customer_id", how="left")
retail_df = retail_df.withColumn("new", F.when(F.col("yyyymm") == F.col("users"), True).otherwise(False))
display(retail_df)
```

```python
n_user_count = retail_df.groupby("yyyymm", "new").agg(F.countDistinct("customer_id").alias("Active Users")).orderBy("yyyymm")
user_pivot = n_user_count.groupBy("yyyymm").pivot("new").agg(F.first("Active Users"))
display(user_pivot)
```

## Finding RFM

```python
R = retail_df.groupby("customer_id").agg(F.max("invoice_date").alias("invoice_date"))
R = R.withColumn("recency",F.datediff(F.current_date(), F.col("invoice_date")))
F_df = retail_df.groupby("customer_id").agg(F.countDistinct("invoice_no").alias("Frequency"))
M = retail_df.groupby("customer_id").agg(F.round(F.sum("total"),2).alias("Monetary"))
RFM = R.join(F_df, on = "customer_id", how = "left").join(M, on = "customer_id", how = "left")
RFM = RFM.drop("invoice_date").orderBy("customer_id")
display(RFM)
```

# RFM Segmentation

```python
recency_discretizer = QuantileDiscretizer(numBuckets=5, inputCol="Recency", outputCol="RecencyBucket")
frequency_discretizer = QuantileDiscretizer(numBuckets=5, inputCol="Frequency", outputCol="FrequencyBucket")
monetary_discretizer = QuantileDiscretizer(numBuckets=5,inputCol="Monetary",outputCol="MonetaryBucket")

RFM = recency_discretizer.fit(RFM).transform(RFM)
RFM = frequency_discretizer.fit(RFM).transform(RFM)
RFM = monetary_discretizer.fit(RFM).transform(RFM)

RFM = (RFM.withColumn("RecencyScore", 5 - col("RecencyBucket"))
    .withColumn("FrequencyScore", col("FrequencyBucket") + 1)
    .withColumn("MonetaryScore", col("MonetaryBucket") + 1))

RFM = RFM.drop("RecencyBucket", "FrequencyBucket", "MonetaryBucket")

display(RFM)
```

```python
#calculation of the RFM score
RFM = RFM.withColumn("RFM_SCORE",concat_ws("",col("RecencyScore").cast("int"),
                                           col("FrequencyScore").cast("int"),
                                           col("MonetaryScore").cast("int")))
display(RFM)
```

```python
RFM = RFM.withColumn("RF_Score",concat_ws("",col("RecencyScore").cast("int"),col("FrequencyScore").cast("int")))

RFM = RFM.withColumn(
    "Segment",
    when(col("RF_SCORE").rlike("^[1-2][1-2]$"), "Hibernating")
    .when(col("RF_SCORE").rlike("^[1-2][3-4]$"), "At Risk")
    .when(col("RF_SCORE").rlike("^[1-2]5$"), "Can't Lose")
    .when(col("RF_SCORE").rlike("^3[1-2]$"), "About to Sleep")
    .when(col("RF_SCORE").rlike("^33$"), "Need Attention")
    .when(col("RF_SCORE").rlike("^[3-4][4-5]$"), "Loyal Customers")
    .when(col("RF_SCORE").rlike("^41$"), "Promising")
    .when(col("RF_SCORE").rlike("^51$"), "New Customers")
    .when(col("RF_SCORE").rlike("^[4-5][2-3]$"), "Potential Loyalists")
    .when(col("RF_SCORE").rlike("^5[4-5]$"), "Champions")
    .otherwise("Others")
)

display(RFM)
```

```python
RFM_table = RFM.groupBy("Segment").agg(
    F.count("*").alias("count"),
    F.mean("Recency").alias("Recency_mean"),
    F.mean("Frequency").alias("Frequency_mean"),
    F.mean("Monetary").alias("Monetary_mean")
).orderBy(F.col("segment").asc())
display(RFM_table)
```
